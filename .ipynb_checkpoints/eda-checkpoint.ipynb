```python
import pandas as pd
df = pd.read_csv('data\delivery_data.csv')
# Quick look
print(df.head())
print(df.info())
```

```python
print(df.info())           # data types & nulls
print(df.describe())       # summary for numerical columns
print(df.isnull().sum())   # missing values check
```

```python
# Overwrite the original label column
df['label'] = df['label'].apply(lambda x: 0 if x in [0, -1] else 1)

# Check distribution
print(df['label'].value_counts(normalize=True))
```

```python
print(df.describe()) 
print(df.info())
print(df['label'].value_counts())
```

```python
import seaborn as sns
sns.countplot(x='label', data=df)
```
